```python
import time
from utils.coin_env import coin_game
```

```python
env = coin_game()
env.reset()  # Reset game ke kondisi awal (30 koin, giliran Player 1)
print(f"the current state is state={env.state}") 

# --- SIKLUS UTAMA SIMULASI PERMAINAN (MAIN GAME LOOP) ---
while True:
    action = env.sample()  # 1. Pilih aksi acak legal dari sisa koin
    time.sleep(1)          # Jeda 1 detik agar simulasi nyaman diamati di terminal
    print(f"Player {env.turn} has chosen {action}")    
    
    state, reward, done, info = env.step(action)  # 2. Transisi State ke environment
    print(f"the current state is state={env.state}") 
    
    # 3. Evaluasi Kondisi Akhir Permainan (Game Over)
    if done:
        if reward == 1:
            print("Player 1 has won!")   # Reward = 1 mengindikasikan Player 1 menang
        else:
            print("Player 2 has won!")   # Reward = -1 mengindikasikan Player 2 menang
            
        break  # Hentikan loop simulasi
```
